```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris

# Load the Iris dataset
iris = load_iris(as_frame=True)
df = iris.frame

# Replace numeric labels with species names
df["species"] = df["target"].map(
    dict(enumerate(iris.target_names))
)

# Display the first rows
print("First 5 rows:")
print(df.head())

# Dataset overview
print("\nDataset shape:", df.shape)
print("\nDataset information:")
df.info()

# Check missing values
print("\nMissing values:")
print(df.isnull().sum())

# Summary statistics
print("\nSummary statistics:")
print(df.drop(columns=["target"]).describe())
```

```python

# Count samples for each species
print(df["species"].value_counts())

# Average petal length by species
print(
    df.groupby("species")["petal length (cm)"]
    .mean()
)

# Plot 1: Species distribution
sns.countplot(data=df, x="species")
plt.title("Number of Samples per Species")
plt.xlabel("Species")
plt.ylabel("Count")
plt.tight_layout()
plt.show()

# Plot 2: Petal length comparison
sns.boxplot(
    data=df,
    x="species",
    y="petal length (cm)"
)
plt.title("Petal Length by Species")
plt.tight_layout()
plt.show()

# Plot 3: Relationship between petal features
sns.scatterplot(
    data=df,
    x="petal length (cm)",
    y="petal width (cm)",
    hue="species"
)
plt.title("Petal Length vs Petal Width")
plt.tight_layout()
plt.show()
```
